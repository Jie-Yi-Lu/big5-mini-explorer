```python
import pandas as pd
import matplotlib.pyplot as plt

# --- Step 1: Load ---
# Defence: specify encoding explicitly; errors='replace' avoids crash on bad bytes
df = pd.read_csv('../data/raw/BIG5/data.csv', sep='\t', encoding='utf-8', encoding_errors='replace')
total = len(df)

# Defence: confirm separator worked — expect many columns, not just 1
assert df.shape[1] > 2, f'Separator may be wrong — only {df.shape[1]} column(s) found'

# --- Step 2: Inspect dtype ---
# Defence: coerce non-numeric age values to NaN instead of raising
df['age'] = pd.to_numeric(df['age'], errors='coerce')
n_missing = df['age'].isna().sum()
if n_missing:
    print(f'Warning: {n_missing} non-numeric age value(s) set to NaN and will be dropped')

# --- Step 3: Filter ---
df_filtered = df[(df['age'] >= 13) & (df['age'] <= 80)]
kept = len(df_filtered)
print(f'Total rows loaded : {total}')
print(f'Rows kept (13-80) : {kept}  (dropped {total - kept})')

# --- Step 4: Plot ---
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(df_filtered['age'], bins=30, color='#4C72B0', edgecolor='white', linewidth=0.4)
ax.set_xlabel('Age (years)')
ax.set_ylabel('Count')
ax.set_title(f'BIG5 – Age Distribution (n={kept})')
fig.tight_layout()

# --- Step 5: Save ---
output = '../reports/age_distribution.png'
fig.savefig(output, dpi=150)
print(f'Figure saved to   : {output}')
plt.show()
```
